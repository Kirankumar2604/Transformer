```python
import math
```

```python
# Step 1: Input sentence
text = "what is the main purpose of transformer architecture"
```

```python
# Step 2: Tokenization
tokens = text.split()
```

```python
# Step 3: Create illustrative embedding vectors
embedding_dim = 4

embeddings = []

for i, token in enumerate(tokens):
    vector = [
        round((i + 1) * 0.1, 2),
        round((i + 1) * 0.2, 2),
        round((i + 1) * 0.3, 2),
        round((i + 1) * 0.4, 2)
    ]
    embeddings.append(vector)
```

```python
# Step 4: Calculate sinusoidal positional encoding
def positional_encoding(position, dimension):
    encoding = []

    for i in range(dimension):
        angle = position / (
            10000 ** (2 * (i // 2) / dimension)
        )

        if i % 2 == 0:
            value = math.sin(angle)
        else:
            value = math.cos(angle)

        encoding.append(value)

    return encoding
```

```python
# Step 5: Add positional encoding to embeddings
final_vectors = []

for position, embedding in enumerate(embeddings):
    pos_vector = positional_encoding(
        position, embedding_dim
    )

    combined = [
        embedding[i] + pos_vector[i]
        for i in range(embedding_dim)
    ]

    final_vectors.append(combined)
```

```python
# Step 6: Display results
for token, embedding, result in zip(
    tokens, embeddings, final_vectors
):
    print("\nToken:", token)
    print("Embedding:", embedding)
    print(
        "Position encoding:",
        [round(x, 4) for x in positional_encoding(
            tokens.index(token), embedding_dim
        )]
    )
    print(
        "Final vector:",
        [round(x, 4) for x in result]
    )
```
