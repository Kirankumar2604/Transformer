```python

import math

# ==========================================
# STEP 1: Original input sentence
# ==========================================

text = "what is the main purpose of transformer architecture"

tokens = text.split()

print("Input tokens:", tokens)


# ==========================================
# STEP 2: Simulate final hidden representation
# ==========================================

# In a real Transformer, this comes from the
# final Transformer layer for the last input token.
# These values are illustrative.

hidden = [0.5, 0.8, 0.3, 0.9]


# ==========================================
# STEP 3: Define a small output vocabulary
# ==========================================

candidate_tokens = [
    ".", "because", "and", "banana","?"
]

# One row per candidate token.
# Four input dimensions -> four candidate scores.

W_output = [
    [1.0, 0.5, 0.0, 0.0],   # "."
    [0.0, 1.0, 0.5, 0.5],   # "because"
    [0.5, 0.0, 0.5, 0.0],   # "and"
    [-0.5, 0.0, -0.5, 0.0]  # "banana"
]

bias = [0.0, 0.0, 0.0, 0.0]


# ==========================================
# STEP 4: Calculate logits
# ==========================================

logits = []

for row, b in zip(W_output, bias):
    score = sum(
        hidden[i] * row[i]
        for i in range(len(hidden))
    ) + b

    logits.append(score)

print("\nCandidate tokens:")
print(candidate_tokens)

print("\nLogits:")
for token, score in zip(candidate_tokens, logits):
    print(token, round(score, 4))


# ==========================================
# STEP 5: Softmax
# ==========================================

def softmax(values):
    # Numerical stability
    max_value = max(values)

    exp_values = [
        math.exp(value - max_value)
        for value in values
    ]

    total = sum(exp_values)

    return [
        value / total
        for value in exp_values
    ]


probabilities = softmax(logits)

print("\nProbabilities:")

for token, probability in zip(
    candidate_tokens, probabilities
):
    print(
        token,
        round(probability, 4),
        f"({probability * 100:.2f}%)"
    )


# ==========================================
# STEP 6: Select the highest-probability token
# ==========================================

best_index = probabilities.index(max(probabilities))

predicted_token = candidate_tokens[best_index]

print("\nPredicted next token:")
print(predicted_token)


# ==========================================
# STEP 7: Append predicted token
# ==========================================

generated_text = text + predicted_token

print("\nGenerated text:")
print(generated_text)
```
