# Input sentence

```python
text = "what is the main purpose of transformer architecture"

print("Input sentence:")
print(text)
```

## 2. Word-Level Tokenization

```python
word_tokens = text.split()

print("1. Word-Level Tokenization:")
print(word_tokens)
print("Token count:", len(word_tokens))
```

## 2. Character-Level Tokenization

```python
character_tokens = list(text)

print("\n2. Character-Level Tokenization:")
print(character_tokens)
print("Character count:", len(character_tokens))
```

## 3. Subword-Level Tokenization (Illustrative)
## Actual subword splitting depends on the tokenizer.

```python
subword_tokens = [
    "what", "is", "the", "main", "purpose",
    "of", "transform", "er", "architecture"
]

print("\n3. Subword-Level Tokenization (Illustrative):")
print(subword_tokens)
print("Token count:", len(subword_tokens))
```

## 4. Byte-Level Representation

```python
byte_values = list(text.encode("utf-8"))

print("\n4. UTF-8 Byte Representation:")
print(byte_values)
print("Byte count:", len(byte_values))
```

## 5. Sentence-Level Tokenization

```python
sentence_tokens = [text]

print("\n5. Sentence-Level Tokenization:")
print(sentence_tokens)
print("Sentence count:", len(sentence_tokens))
```
