```python
# ==========================================
# STEP 1: Input sentence
# ==========================================

text = "what is the main purpose of transformer architecture"
tokens = text.split()

print("Tokens:", tokens)
```

```python
# ==========================================
# STEP 2: Create illustrative input vectors
# ==========================================

embedding_dim = 4
X = []

for i, token in enumerate(tokens):
    vector = [
        (i + 1) * 0.1,
        (i + 1) * 0.2,
        (i + 1) * 0.3,
        (i + 1) * 0.4
    ]
    X.append(vector)
```

```python
# ==========================================
# STEP 3: Matrix-vector multiplication
# ==========================================

def linear(vector, weights, bias):
    output = []

    for j in range(len(weights[0])):
        value = bias[j]

        for i in range(len(vector)):
            value += vector[i] * weights[i][j]

        output.append(value)

    return output
```

```python
# ==========================================
# STEP 4: Define FFN parameters
# ==========================================

# First layer: 4 input dimensions -> 8 hidden dimensions
W1 = [
    [0.1, 0.2, 0.1, 0.3, 0.2, 0.1, 0.2, 0.1],
    [0.2, 0.1, 0.3, 0.1, 0.1, 0.2, 0.1, 0.3],
    [0.1, 0.3, 0.2, 0.1, 0.3, 0.1, 0.2, 0.2],
    [0.3, 0.1, 0.1, 0.2, 0.1, 0.3, 0.1, 0.2]
]

b1 = [0.0] * 8

# Second layer: 8 hidden dimensions -> 4 output dimensions
W2 = [
    [0.1, 0.2, 0.1, 0.2],
    [0.2, 0.1, 0.2, 0.1],
    [0.1, 0.1, 0.2, 0.3],
    [0.2, 0.3, 0.1, 0.1],
    [0.1, 0.2, 0.3, 0.1],
    [0.3, 0.1, 0.1, 0.2],
    [0.2, 0.1, 0.2, 0.1],
    [0.1, 0.2, 0.1, 0.3]
]

b2 = [0.0] * 4
```

```python
# ==========================================
# STEP 5: ReLU activation
# ==========================================

def relu(vector):
    return [max(0.0, x) for x in vector]
```

```python
# ==========================================
# STEP 6: Apply FFN to each token
# ==========================================

ffn_outputs = []

for token, vector in zip(tokens, X):

    # First linear transformation: 4 -> 8
    hidden = linear(vector, W1, b1)

    # Nonlinear activation
    activated = relu(hidden)

    # Second linear transformation: 8 -> 4
    output = linear(activated, W2, b2)

    ffn_outputs.append(output)

    print("\nToken:", token)
    print("Input:", [round(x, 4) for x in vector])
    print("Hidden:", [round(x, 4) for x in hidden])
    print("After ReLU:", [round(x, 4) for x in activated])
    print("FFN output:", [round(x, 4) for x in output])
```

```python
# ==========================================
# STEP 7: Verify dimensions
# ==========================================

print("\nNumber of tokens:", len(ffn_outputs))
print("Output dimensions per token:", len(ffn_outputs[0]))
```
