```python

import math

# ==========================================
# STEP 1: Input sentence
# ==========================================

text = "what is the main purpose of transformer architecture"
tokens = text.split()

print("Input tokens:", tokens)
```

```python
# ==========================================
# STEP 2: Create illustrative input vectors
# ==========================================

embedding_dim = 4
X = []

for i, token in enumerate(tokens):
    vector = [
        (i + 1) * 0.1,
        (i + 1) * 0.2,
        (i + 1) * 0.3,
        (i + 1) * 0.4
    ]
    X.append(vector)
```

```python
# ==========================================
# STEP 3: Simulate attention output
# ==========================================

# In the previous step, this would be the
# output of multi-head attention.
# Here, fixed values simplify the demonstration.

attention_output = []

for vector in X:
    attention_output.append([
        value * 0.1 for value in vector
    ])
```

```python
# ==========================================
# STEP 4: Residual connection (ADD)
# ==========================================

residual_output = []

for original, attended in zip(X, attention_output):
    combined = [
        original[i] + attended[i]
        for i in range(embedding_dim)
    ]
    residual_output.append(combined)
```

```python
# ==========================================
# STEP 5: Layer normalization
# ==========================================

def layer_norm(vector, epsilon=1e-5):
    # Calculate mean
    mean = sum(vector) / len(vector)

    # Calculate population variance
    variance = sum(
        (value - mean) ** 2
        for value in vector
    ) / len(vector)

    # Normalize
    normalized = [
        (value - mean) /
        math.sqrt(variance + epsilon)
        for value in vector
    ]

    # Simplified learnable parameters:
    # gamma = 1, beta = 0
    gamma = [1.0] * len(vector)
    beta = [0.0] * len(vector)

    output = [
        normalized[i] * gamma[i] + beta[i]
        for i in range(len(vector))
    ]

    return output


normalized_output = [
    layer_norm(vector)
    for vector in residual_output
]
```

```python
# ==========================================
# STEP 6: Display results
# ==========================================

for token, original, attended, added, normalized in zip(
    tokens,
    X,
    attention_output,
    residual_output,
    normalized_output
):
    print("\nToken:", token)
    print("Original input:", original)
    print("Attention output:", attended)
    print("After residual addition:", added)
    print(
        "After layer normalization:",
        [round(value, 4) for value in normalized]
    )
```
