* ==========================================
* STEP 1: Input sentence
* ==========================================

```python
text = "what is the main purpose of transformer architecture"

print("Input sentence:")
print(text)
```

* ==========================================
* STEP 2: Tokenization
* ==========================================

```python
tokens = text.split()

print("\nTokens:")
print(tokens)

print("\nNumber of tokens:", len(tokens))
```

* ==========================================
* STEP 3: Build vocabulary
* ==========================================

```python
# Assign a unique integer ID to each word.

vocabulary = {}

for token in tokens:
    if token not in vocabulary:
        vocabulary[token] = len(vocabulary)

print("\nVocabulary:")
print(vocabulary)
```

```python
# ==========================================
# STEP 4: Convert tokens into IDs
# ==========================================

token_ids = []

for token in tokens:
    token_ids.append(vocabulary[token])

print("\nToken IDs:")
print(token_ids)
```

```python
# ==========================================
# STEP 5: Create an embedding matrix
# ==========================================

# For learning purposes, use 4 dimensions.
# Real embedding values are usually learned
# during neural network training.

embedding_dim = 4
vocab_size = len(vocabulary)

embedding_matrix = []

for i in range(vocab_size):
    # Illustrative values, not trained embeddings.
    vector = [
        round((i + 1) * 0.1, 2),
        round((i + 1) * 0.2, 2),
        round((i + 1) * 0.3, 2),
        round((i + 1) * 0.4, 2)
    ]

    embedding_matrix.append(vector)

print("\nEmbedding Matrix:")

for word, word_id in vocabulary.items():
    print(word, "->", embedding_matrix[word_id])
```

* ==========================================
* STEP 6: Embedding lookup
* ==========================================

```python
embedded_vectors = []

for token_id in token_ids:
    vector = embedding_matrix[token_id]
    embedded_vectors.append(vector)

print("\nFinal Embedding Vectors:")

for token, vector in zip(tokens, embedded_vectors):
    print(token, "->", vector)
```

* ==========================================
* STEP 7: Display dimensions
* ==========================================

```python


print("\nEmbedding matrix shape:")
print(vocab_size, "x", embedding_dim)

print("\nOutput shape:")
print(len(embedded_vectors), "x", embedding_dim)
```
